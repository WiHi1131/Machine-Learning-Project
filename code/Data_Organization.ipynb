```python
import json
import os
from collections import defaultdict

# Replace 'matches.json' with the path to your original JSON file.
input_file = 'match_data_set13.json'

with open(input_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

# Dictionary where each key is a tft_game_type and each value is a list of entries of that type
grouped_data = defaultdict(list)

for entry in data:
    game_type = entry["info"]["tft_game_type"]
    grouped_data[game_type].append(entry)

# Create a new JSON file for each tft_game_type
for game_type, entries in grouped_data.items():
    output_filename = f"{game_type}_entries.json"
    with open(output_filename, 'w', encoding='utf-8') as out:
        json.dump(entries, out, indent=4, ensure_ascii=False)
    print(f"Created file: {output_filename} with {len(entries)} entries.")
```

```python
import json
import csv

# The path to your JSON file containing standard entries
input_json = "SET_13_STANDARD_MATCHES.json"
# The path to your desired CSV output
output_csv = "Q1_GOLD_LEFT_V_PLACEMENT.csv"

# Load the data from the JSON file
with open(input_json, "r", encoding="utf-8") as f:
    data = json.load(f)

# Open a CSV file in write mode
with open(output_csv, "w", newline="", encoding="utf-8") as csvfile:
    writer = csv.writer(csvfile)

    # Write header row
    writer.writerow(["match_id", "puuid", "gold_left", "placement"])

    # Loop through every entry in the JSON
    for entry in data:
        match_id = entry["metadata"]["match_id"]

        # Loop through each participant in the current match
        for participant in entry["info"]["participants"]:
            puuid = participant["puuid"]
            placement = participant["placement"]
            gold_left = participant.get("gold_left", None)  # .get() avoids errors if key is missing

            # Write a row for each participant
            writer.writerow([match_id, puuid, gold_left, placement])

print(f"CSV file '{output_csv}' has been created successfully.")
```

```python
import json
import csv

# Input and output file names
input_file = "SET_13_STANDARD_MATCHES.JSON"
output_csv = "Q2_TRAITS_V_PLACEMENT.csv"

# Read the JSON file
with open(input_file, "r", encoding="utf-8") as f:
    data = json.load(f)

# Open the CSV file in write mode
with open(output_csv, "w", newline="", encoding="utf-8") as csvfile:
    writer = csv.writer(csvfile)

    # Write a header row
    writer.writerow(["match_id", "puuid", "trait_name", "num_units", "tier_current", "tier_total", "placement"])

    # Loop over every match in the JSON
    for match_entry in data:
        match_id = match_entry["metadata"]["match_id"]

        # Loop through each participant in this match
        for participant in match_entry["info"]["participants"]:
            puuid = participant["puuid"]
            placement = participant["placement"]

            # For each trait in this participant
            for trait in participant["traits"]:
                trait_name = trait["name"]
                num_units = trait["num_units"]
                tier_current = trait["tier_current"]
                tier_total = trait["tier_total"]

                # Write one CSV row per trait
                writer.writerow([
                    match_id, 
                    puuid, 
                    trait_name, 
                    num_units, 
                    tier_current, 
                    tier_total,
                    placement
                ])

print(f"CSV file '{output_csv}' has been created with trait data.")
```

```python
import json
import csv

# Input and output file names
input_file = "SET_13_STANDARD_MATCHES.JSON"
output_csv = "PRE_PCA.csv"

# Read the JSON file
with open(input_file, "r", encoding="utf-8") as f:
    data = json.load(f)

# Open the CSV file in write mode
with open(output_csv, "w", newline="", encoding="utf-8") as csvfile:
    writer = csv.writer(csvfile)

    # Write a header row
    writer.writerow(["match_id", "game_length", "puuid", "gold_left", "last_round", "players_eliminated", "time_eliminated", "total_damage_to_players", "placement"])

    # Loop over every match in the JSON
    for match_entry in data:
        match_id = match_entry["metadata"]["match_id"]
        game_length = match_entry["info"]["game_length"]

        # Loop through each participant in this match
        for participant in match_entry["info"]["participants"]:
            puuid = participant["puuid"]
            placement = participant["placement"]
            gold_left = participant["gold_left"]
            last_round = participant["last_round"]
            players_eliminated = participant["players_eliminated"]
            time_eliminated = participant["time_eliminated"]
            total_damage_to_players = participant["total_damage_to_players"]
            # Write one CSV row per trait
            writer.writerow([
                match_id, 
                game_length,
                puuid, 
                gold_left,
                last_round,
                players_eliminated,
                time_eliminated,
                total_damage_to_players,
                placement
            ])

print(f"CSV file '{output_csv}' has been created with trait data.")
```
